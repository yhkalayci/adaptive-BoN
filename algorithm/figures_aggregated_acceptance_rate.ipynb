```python
import numpy as np
import matplotlib.pyplot as plt

def plot_aggregate_acceptance_rate(data, figsize=(14, 5), figure_title=""):
    aggregate = data["aggregate"]
    x = np.array([item["target_acceptance_rate"] for item in aggregate], dtype=float)
    
    save_median = np.array([item["save_actual_median"] for item in aggregate], dtype=float)
    save_p25 = np.array([item["save_actual_p25"] for item in aggregate], dtype=float)
    save_p75 = np.array([item["save_actual_p75"] for item in aggregate], dtype=float)
    
    achieved_median = np.array([item["achieved_acceptance_rate_median"] for item in aggregate], dtype=float)
    achieved_p25 = np.array([item["achieved_acceptance_rate_p25"] for item in aggregate], dtype=float)
    achieved_p75 = np.array([item["achieved_acceptance_rate_p75"] for item in aggregate], dtype=float)
    
    fig, axes = plt.subplots(1, 2, figsize=figsize)
    
    line_color = "#1f77b4"
    
    axes[0].plot(x, save_median, "s-", linewidth=2, markersize=6, color=line_color)
    axes[0].fill_between(x, save_p25, save_p75, color=line_color, alpha=0.25)
    axes[0].set_title("Target Acceptance Rate vs Save Ratio", fontsize=13)
    axes[0].set_xlabel("Target Acceptance Rate", fontsize=12)
    axes[0].set_ylabel("Save Ratio (Actual)", fontsize=12)
    axes[0].grid(True, alpha=0.3, linestyle="-", linewidth=0.5)
    axes[0].set_facecolor("#f8f8f8")
    
    axes[1].plot(x, achieved_median, "s-", linewidth=2, markersize=6, color=line_color)
    axes[1].fill_between(x, achieved_p25, achieved_p75, color=line_color, alpha=0.25)
    axes[1].plot([x.min(), x.max()], [x.min(), x.max()], "k--", alpha=0.3, linewidth=1)
    axes[1].set_title("Target Acceptance Rate vs Achieved Acceptance Rate", fontsize=13)
    axes[1].set_xlabel("Target Acceptance Rate", fontsize=12)
    axes[1].set_ylabel("Achieved Acceptance Rate", fontsize=12)
    axes[1].grid(True, alpha=0.3, linestyle="-", linewidth=0.5)
    axes[1].set_facecolor("#f8f8f8")
    
    if figure_title:
        fig.suptitle(figure_title, fontsize=14, y=0.98, fontweight="bold")
    fig.patch.set_facecolor("white")
    plt.tight_layout()
    return fig, axes
```

```python
plot_aggregate_acceptance_rate(data, figure_title=f"{llm_mapping[llm]} | {rm_mapping[rm]}")
```
